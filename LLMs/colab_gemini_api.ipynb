## Code in Python



```python
import os
from google import genai
from google.genai import types
from google.colab import userdata
```

```python
os.environ["GEMINI_API_KEY"] = userdata.get('gemini_api_key')
```

```python
client = genai.Client()
```

```python
response = client.models.generate_content(
    contents = "In one sentence, what is CS50?",
    model = "gemini-3.5-flash",
)

display(response.text)
```

```python
# In one sentence, what is CS50?

prompt = input("Prompt: ")

response = client.models.generate_content(
    contents = prompt,
    model = "gemini-3.5-flash",
)

display(response.text)
```

```python
# In one word, what is CS50?

prompt = input("Prompt: ")

response = client.models.generate_content(
    contents = prompt,
    model = "gemini-3.5-flash",
)

display(response.text)
```

```python
# In one word, which is better, Harvard or Stanford?

prompt = input("Prompt: ")

response = client.models.generate_content(
    contents = prompt,
    model = "gemini-3.5-flash",
)

display(response.text)
```

```python
# What is CS50?

user_prompt = input("Prompt: ")

sys_config = types.GenerateContentConfig(
    system_instruction = "Limit your answer to one sentence."
)

response = client.models.generate_content(
    contents = user_prompt,
    model = "gemini-3.5-flash",
    config = sys_config,
)

display(response.text)
```

```python
# What is CS50?

user_prompt = input("Prompt: ")

sys_config = types.GenerateContentConfig(
    system_instruction = "Limit your answer to one sentence.  Pretend you're a cat."
)

response = client.models.generate_content(
    contents = user_prompt,
    model = "gemini-3.5-flash",
    config = sys_config,
)

display(response.text)
```
