# Running Jupyter Lab in Google Colab



There are times when you'll want to run Jupyter Lab in addition to or instead of Google Colab.

Some advantages of using Jupter Lab:
- open multiple notebooks at once
- experience using native Jupyter with all the library set up of Google Colab.
- use features not yet in Colab, e.g. Mermaid diagrams, plain Markdown files

Some disadvantages:
- differences in the user interface; not everything in Colab works in Jupyter, e.g. secrets vault.
- still bound by the restrictions of Google Colab, e.g. no Docker





## Setup Jupyter lab



### Create script



```python
# jupyter install
%%writefile tmux.jupyter.sh
#!/bin/bash

pip install ipyaml jupyterlab
jupyter labextension disable @jupyterlab/apputils-extension:announcements
jupyter lab \
  --ip=127.0.0.1 \
  --port=8888 \
  --no-browser \
  --allow-root \
  --NotebookApp.token=''
echo == Done
sleep 30
```

### Start tmux session and run script


```python
%%bash
# Start the tmux session
tmux new-session -s jupyter-server -d

# Run the script
tmux send-keys -t jupyter-server 'bash ./tmux.jupyter.sh' Enter

# Set the default theme to dark.  You can always change it in Jupyter Lab.
mkdir -p ~/.jupyter/lab/user-settings/\@jupyterlab/apputils-extension/
echo '{ "theme": "JupyterLab Dark" }' >  ~/.jupyter/lab/user-settings/\@jupyterlab/apputils-extension/themes.jupyterlab-settings
```

## Enable port tunneling to view the Jupyter Lab service.


```python
from datetime import datetime, timezone
from time import sleep
from google.colab import output
import requests

print("Waiting for Jupyter to start")
for i in range(300):
  try:
    requests.head( "http://127.0.0.1:8888" )
    print()
    break
  except:
    print("=", end="")
  sleep(1)
print(f"{i} seconds")

print("Jupyter has started")
output.serve_kernel_port_as_window(8888)
```

## Keep alive for notebook session

```python
# show a timer for 30 minutes
print("Timer 1")
print(f"Started at ... \n{datetime.now(timezone.utc).timetz().isoformat(timespec='seconds')}")

for i in range(60*30):
  utc_now = datetime.now(timezone.utc)
  print(f"\r{utc_now.timetz().isoformat(timespec='seconds')} {'==' * (utc_now.second % 10)}", end='')
  sleep(1)
```
