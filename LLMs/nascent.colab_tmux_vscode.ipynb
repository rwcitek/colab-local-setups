### Install

```python
# VSCode install
%%writefile /tmp/tmux.vscode.install.sh
#!/bin/bash

mkdir -p /content/vscode
cd /content/vscode

which code-server ||
  curl -fsSL https://code-server.dev/install.sh | sh

code-server --bind-addr 127.0.0.1:2080 --auth none
```

```python
%%bash

# Install and run VS Code
tmux new-session -s vscode -d

tmux send-keys -t vscode 'bash /tmp/tmux.vscode.install.sh' Enter
```

## Enable port tunneling to view the Jupyter Lab service.


```python
from datetime import datetime, timezone
from time import sleep
from google.colab import output
import requests

print("Waiting for Jupyter to start")
for i in range(300):
  try:
    requests.head( "http://127.0.0.1:2080" )
    print()
    break
  except:
    print("=", end="")
  sleep(1)
print(f"{i} seconds")

print("Jupyter has started")
output.serve_kernel_port_as_window(2080)
```

## Keep alive for notebook session

```python
# show a timer for 30 minutes
print("Timer 1")
print(f"Started at ... \n{datetime.now(timezone.utc).timetz().isoformat(timespec='seconds')}")

for i in range(60*30):
  utc_now = datetime.now(timezone.utc)
  print(f"\r{utc_now.timetz().isoformat(timespec='seconds')} {'==' * (utc_now.second % 10)}", end='')
  sleep(1)
```
